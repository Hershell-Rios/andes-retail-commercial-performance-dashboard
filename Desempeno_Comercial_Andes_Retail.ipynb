### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

**S (Situación):** Contexto del negocio
- Los ingresos del año 2024 (3 mill.) con 29 mil unidades vendidas, fueron muy similares a las del año 2025 (3 mill.) con 28 mil unidades vendidas. Eso nos indica que no hubo un crecimiento, notamos un estancamiento.

**C (Complicación):**  
- Si segmentamos los ingresos por tipo de cliente, 2 tipo de cliente concentran el 90% de los ingresos (Premium con 47% y Estándar con 43%) considerando el periodo de los dos años. Si comparamos el ingreso entre los tres países en los que se tiene participación (Chile, Colombia y Perú), Perú concentra el 39.02% de los ingresos totales, y Chile concentra el 36.65%, dejando a Colombia con una participación de 24.33%, la más baja de todas; en el mismo perido de los dos años.

**Q (Pregunta):**   
- Si tenemos dos tipos de cliente que lideran los ingresos. ¿El tercer tipo de cliente tiene alguna categoría de producto en la que sobresalga o todas las categorías se comportan semejante?

**A (Respuesta):**  
- En general todo el segmente Económico tiene un desempeño muy bajo, con solo el 10% de participación en los ingresos totales, como en todas las categorías (Hogar, Deportes, Electrónica, Ropa). Tal vez haya que replantear deshacerse del segmento de cliente Económico.

---

#### 🔎 Vista Detalle

**S (Situación):**  
- Los ingresos detallados por región dentro de los paises donde se tiene participación tienen un comportamiento muy similar. Perú en el año 2024 registro 1.14 mill., en el año 2025 registro 1.02 mill. Chile en el año 2024 registro 1.04 mill., en el año 2025 registro 0.99 mill. Colombia en el año 2024 registro 0.69 mill, en el año 2025 registro 0.66 mill. Siendo el país que menos ingresos genero.

**C (Complicación):**  
- Analizando los patrones temporales en los años 2024 y 2025, confirmamos que Colombia tiene un desempeño negativo, sin cumplir en ambos años con el mínimo esperado. Chile tiene un desempeño en general bueno, pero tiene algunos meses que tiene un registro negativo en ingresos. Perú tiene un desempeño muy positivo a lo largo de los dos años, superando con creces el nivel mínimo esperado.

**Q (Pregunta):**  
- ¿Colombia tiene alguna temporada donde registre ventas positivas o todas temporadas se comporta de la misma manera?

**A (Respuesta):**  
- Incluso haciendo un análisis en las estaciones registradas (Primavera, Verano, Otoño e Invierno), se confirma que en ninguna estación se cumplio el minimo esperado en el nivel de ingresos en Colombia. Chile tiene algunos meses que tiene un nivel de ingresos bajos, pero se recupera en los meses siguientes. Y Perú es la que tiene un desempeño positivo en ambos años. La recomndación sería analizar las estrategías que se están aplicando en Colombia, o revisar si hay factores que influyan en su desempeño. O si los prodcutos que se están vendiendo no son los adecuados para el país.
